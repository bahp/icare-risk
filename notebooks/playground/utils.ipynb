```python
import pandas as pd
from pathlib import Path

def load_and_deduplicate(file_path: str | Path, subset: list[str] | None = None, keep: str = 'first') -> pd.DataFrame:
    """Loads a CSV file into a pandas DataFrame and removes duplicate rows.

    Parameters
    ----------
    file_path: Path to the CSV file.
    subset: Optional list of column names to consider for identifying duplicates. If None, uses all columns.
    keep: Which duplicates to keep ('first', 'last', or False to drop all duplicates).
    """
    df = pd.read_csv(file_path)
    initial_count = len(df)

    # Drop duplicates
    df_clean = df.drop_duplicates(subset=subset, keep=keep)
    removed_count = initial_count - len(df_clean)

    print(f"Loaded {initial_count} rows. Removed {removed_count} duplicates. {len(df_clean)} rows remaining.")
    return df_clean

path = '/app/data/lookups/icare/problems_lookup.csv'
path = '/app/data/lookups/icare/vital_signs_lookup.csv'
df = load_and_deduplicate(path)
df = df.rename(columns={
    'problem_code': 'code',
    'problem_desc': 'name',
})

df.to_csv(path)

print(df.code.nunique())

```
