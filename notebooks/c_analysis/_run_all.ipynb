```python
from pathlib import Path
import nbformat
from nbconvert.preprocessors import ExecutePreprocessor

NOTEBOOK_DIR = Path(".")
OUTPUT_DIR = NOTEBOOK_DIR / "executed"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

pipeline = [
    #"00_data_audit.ipynb",
    #"01_cohort_and_temporal_split.ipynb",
    #"02_clinical_scores.ipynb",
    #"03_train_models.ipynb",
    #"04_generate_predictions.ipynb",
    #"05_model_performance.ipynb",
    #"08_subgroup_analysis.ipynb",
    #"09_feature_importance.ipynb",
    #"10_shap_analysis.ipynb",
    #"11_parsimonious_models.ipynb",
    #"12_sensitivity_analysis.ipynb",
    #"13_threshold_analysis.ipynb",
    #"14_decision_curve_analysis.ipynb",
    #"15_final_model_comparison.ipynb",
]

# Set execution options (timeout in seconds, e.g., 600s per notebook)
executor = ExecutePreprocessor(timeout=600, kernel_name="python3")

for notebook in pipeline:
  input_path = NOTEBOOK_DIR / notebook
  output_path = OUTPUT_DIR / f"executed_{notebook}"
  print(f"\n=== Executing {notebook} via nbconvert ===")

  # 1. Read input notebook
  with open(input_path, "r", encoding="utf-8") as f:
    nb = nbformat.read(f, as_version=4)

  # 2. Execute notebook in its parent directory context
  executor.preprocess(nb, {"metadata": {"path": str(NOTEBOOK_DIR.resolve())}})

  # 3. Save executed output notebook (with outputs & errors saved in cells)
  with open(output_path, "w", encoding="utf-8") as f:
    nbformat.write(nb, f)

print("\nPipeline completed successfully!")
```
