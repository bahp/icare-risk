# 02f — Run the complete pipeline

Execute notebooks 02a–02e in order. This is optional; running the notebooks individually is easier for debugging.

```python
import papermill as pm
from pathlib import Path

NOTEBOOK_DIR = Path(".")
pipeline = [
    "02a_vectorized.ipynb",
    "02b_phenotypes.ipynb",
    "02c_encounters.ipynb",
    "02d_medications.ipynb",
    "02e_microbiology.ipynb",
    "02f_combine.ipynb",
]

for notebook in pipeline:
    notebook_path = NOTEBOOK_DIR / notebook
    output_notebook = NOTEBOOK_DIR / f"executed_{notebook}"
    print(f"\n=== Running {notebook} via Papermill ===")

    pm.execute_notebook(
        input_path=str(notebook_path),
        output_path=str(output_notebook),
        kernel_name="python3",
        cwd=str(NOTEBOOK_DIR.resolve())  # Resolve path issue
    )

print("\nPipeline completed successfully!")
```
